# <a id="5">  5 ) . What is the distribution for the trip duration in minutes? </a>

```python
## Review the statistic for trip duration in minutes
df["duration_min"].describe()
```

```python
# Make visualization to easy get insight

plt.figure(figsize=(10,5), dpi=200)
plt.hist(df['duration_min'], bins=50, color='skyblue', edgecolor='black')
plt.xlabel('Trip Duration (minutes)')
plt.ylabel('Frequency (Number of Trips)')
plt.title("Distribution of Users by Trip Duration (with outliers)")
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
The chart shows the distribution of trip durations for all bike share users.

Most trips (nearly 175,000) are heavily concentrated near the 0–30 minute range. This means the vast majority of users take very short rides, which is typical for bike share systems where people use bikes for quick commutes or short errands.

However, the X-axis extends all the way to 1,400 minutes (over 23 hours). This reveals clear outliers – a small number of trips with extremely long durations that stretch the chart wide open. Because of these outliers, the main cluster of short trips gets squished into a tiny bar on the left, making it hard to see the exact shape of the normal distribution.

Start station longitude and End station longitude also have a strong positive correlation of 0.99. Trips start and end at stations with similar longitudes.

Moderate Correlations:

Weak Correlations:
