```python
def array_reshape(arr, row:int =-1 , cols:int =-1):
    try:
        arr=np.array(arr)
        return arr.reshape(row,cols)
    except (TypeError , ValueError) as e:
        raise ValueError(f"Invalid reshape operation: {e}")
        
```

```python
vec = [1,2,3,4,5,6]
array_reshape(vec , 2, 3)
```
