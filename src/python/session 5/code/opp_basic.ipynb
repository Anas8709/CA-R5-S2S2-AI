```python
class Calc:
    def __init__(self,x:int ,y:int):
        self.x=x
        self.y=y
    def add(self):
        return self.x + self.y  
    def sub(self):
            return self.x - self.y
```

```python
c1=Calc(20 , 30)
c1.sub()
```
