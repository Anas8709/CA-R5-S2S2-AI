```python
class Prime:
    '''A class for checking whether a number is prime.'''
    def __init__(self , num:int):
        self.num = num
    def isPrime(self)->bool:
        '''Check whether the number is prime and return True or False.'''
        for i in range(2, self.num):
            if self.num % i == 0:
                return False
        return True

P1 = Prime(20)
print(P1.isPrime())    
```

```python
class Dividor:
    '''A class for finding the common divisors of two numbers.'''
    
    def __init__(self , num1:int ,num2:int):
        self.num1 = num1
        self.num2 = num2
    
    def dividors(self)->list:
        '''Find and return the common divisors of the two numbers.'''
        
        mn = min(self.num1, self.num2)
        common_dividors = []
        for i in range(1, mn+1):
            if self.num1 % i == 0 and self.num2 % i ==0:
                common_dividors.append(i)
        return common_dividors
    
D1 = Dividor(20 , 10)  
print(D1.dividors())  
```

```python
class Even:
    '''A class for finding the even numbers in a given range.'''
    
    def __init__(self , num:int):
        self.num1 = num
        self.even=[]
     
    
    def even_fun(self)->bool:
        '''Find and return the even numbers in the given range.'''
        
        for i in range(self.num1):
            if i % 2==0:
                self.even.append(i)
        return self.even   
    
E1 = Even(10)  
E1.even_fun()      
```

```python
class Factorial:
    '''A class for calculating the factorial of a number.'''
    
    def fac(self , num:int)->int:
        '''Calculate and return the factorial of the number.'''
        if num == 0:
            return 0
        if num == 1:
            return 1
        return num * self.fac(num -1 )


F1 = Factorial()
print(F1.fac(5))
```

```python
class Reverse:
    '''A class for reversing the order of words in a string.'''
    
    def __init__(self , s:list):
        self.s = s

    def fun(self)->list:
        '''Reverse the order of words in the given string and return it.'''
        lst = self.s.split(" ")
        new_lst = lst[::-1]
        print(new_lst)
        new_str = ""
        for i in new_lst:
            new_str += i + " "
            print(new_str)
        new_str = new_str[:-1]
        return new_lst
    
    
R1 = Reverse("Hello World")   
R1.fun() 
```

```python
class Password_Generator: 
    '''A class for generating random passwords of a specified length.'''
   
    def __init__(self , length:int):
        self.length = length
    
    def generate_password(self)->str:
        '''Generate and return a random password of the specified length.'''
        
        import random
        characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
        password = ""

        for _ in range(self.length):
            password += random.choice(characters)

        return password
    
P1 = Password_Generator(9)  
print(P1.generate_password())  
```

```python
class Frequency_Occurrence:
    '''A class for finding the frequency of occurrence of words.'''
    
    def __init__(self , words:list):
        self.words = words
        
    def get_Occurrence(self)->None:
        '''Find and print the number of times each unique word occurs.'''
        
        unique_words = set(self.words)

        for word in unique_words:
            print(f"{word}:{self.words.count(word)}")  
            
F1 = Frequency_Occurrence(["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"])     
F1.get_Occurrence()      
```
