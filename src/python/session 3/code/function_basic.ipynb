```python
def train_model(model_name , *datasets , **hyperparamemters):
    print(f"Training model: {model_name}")
    print("\n Datasets")
    for dataset in datasets:
        print("_" , dataset)
    print("\n hyperparamemters")
    for key , value in hyperparamemters.items():
        print(f"{key} : {value}")
        
```

```python
train_model(
    "fraide detection" ,
    "train.csv" , 
    "valid.csv" ,
    "test.csv" ,
    learning_rate = 0.01 ,
    batch_size = 32 ,
    epochs = 10 ,
    optimizer = "Adam" ,
    earlystopping = True
)
```
