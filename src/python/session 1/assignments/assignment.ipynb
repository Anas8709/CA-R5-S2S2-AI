```python
#Task 1
Email = "Amit_ml@gmail.edu"

if Email.count("@") < 1 or Email.count(".") < 1:
    print("Invalid email")

print(Email[0 : 7])
print(Email[8 : 13])  

if Email[13 :]  == ".com" :
    print( "Commercial Domain")
elif Email[13 :]  == ".edu" :
    print( "Educational Domain")
else :
    print("Other Domain")
```

```python
#Task 2
vowel = ['A','I','O','E','U']
message =  "emocleW EPGTQ"
message = message.split()
message[0] = "".join(reversed(message[0]))
for x in vowel:
    message[1] = message[1].replace(x ,"")
print(message[0],message[1])
```
