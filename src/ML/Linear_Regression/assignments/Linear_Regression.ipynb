```python
from Class import *
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , mean_absolute_error , r2_score , root_mean_squared_error
```

```python
X = [50, 60, 70, 80, 90] # Features Size of the house
y = [150, 180, 210, 240, 270] #Target Price of the house
X = np.array(X)
y = np.array(y)
```

```python
x_train , x_test , y_train , y_test = train_test_split(
    X , 
    y ,
    train_size= 0.80,
    random_state = 10
)
```

```python
lin_reg = LinearRegressionGD(learning_rate = 0.0001 , n_iters = 100 )
lin_reg.fit(x_train , y_train)
print(f"theta_0 = {lin_reg.theta_0}") # bais
print(f"theta_1 = {lin_reg.theta_1}") # W
y_hat = lin_reg.predict(x_train)
```

```python
predicpredicted_value = lin_reg.predict(70)
print(predicpredicted_value)
```

```python
lin_reg.plot_training(x_train , y_train)
```
